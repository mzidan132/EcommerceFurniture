```python
# Fill missing numerical values with the median
for col in df.select_dtypes(include=['float64', 'int64']).columns:
    df[col].fillna(df[col].median(), inplace=True)

# Fill missing categorical values with the mode
for col in df.select_dtypes(include=['object', 'bool']).columns:
    df[col].fillna(df[col].mode()[0], inplace=True)
```

```python
from sklearn.preprocessing import LabelEncoder

# Label encode the target
label_encoder = LabelEncoder()
df['Current status of microbiota'] = label_encoder.fit_transform(df['Current status of microbiota'])

# One-hot encode all other categorical columns
df = pd.get_dummies(df, drop_first=True)
```

```python
# Avoid division by zero
df['Plant-to-Animal Protein Ratio'] = df['Weekly consumption of plant proteins (portions)'] / (
    df['Weekly consumption of animal proteins (portions)'] + 1e-5)
```

```python
# Bin stress level into categories: Low (1–3), Medium (4–6), High (7–10)
def bin_stress(level):
    if level <= 3:
        return 'Low'
    elif level <= 6:
        return 'Medium'
    else:
        return 'High'

df['Stress Level Category'] = df['Stress level (1-10 scale)'].apply(bin_stress)
df = pd.get_dummies(df, columns=['Stress Level Category'], drop_first=True)
```

Step 5

```python
def evaluate_model(y_true, y_pred, y_proba, model_name="Model", average='macro'):
    import matplotlib.pyplot as plt
    from sklearn.metrics import (
        accuracy_score, precision_score, recall_score, f1_score, roc_auc_score,
        confusion_matrix, ConfusionMatrixDisplay, RocCurveDisplay
    )

    acc = accuracy_score(y_true, y_pred)
    prec = precision_score(y_true, y_pred, average=average)
    rec = recall_score(y_true, y_pred, average=average)
    f1 = f1_score(y_true, y_pred, average=average)

    # Handle AUC-ROC
    try:
        if len(set(y_true)) == 2:
            auc = roc_auc_score(y_true, y_proba[:, 1])
        else:
            auc = roc_auc_score(y_true, y_proba, multi_class='ovr')
    except:
        auc = None

    # Print metrics
    print(f"\n📊 {model_name} Evaluation Metrics:")
    print(f"Accuracy: {acc:.4f}")
    print(f"Precision: {prec:.4f}")
    print(f"Recall: {rec:.4f}")
    print(f"F1-score: {f1:.4f}")
    if auc is not None:
        print(f"AUC-ROC: {auc:.4f}")

    # Confusion Matrix
    cm = confusion_matrix(y_true, y_pred)
    disp = ConfusionMatrixDisplay(confusion_matrix=cm)
    disp.plot(cmap='Blues')
    plt.title(f"{model_name} - Confusion Matrix")
    plt.grid(False)
    plt.show(block=True)

    # ROC Curve
    try:
        RocCurveDisplay.from_predictions(y_true, y_proba, name=model_name)
        plt.title(f"{model_name} - ROC Curve")
        plt.grid(True)
        plt.show()
    except:
        print("ROC curve not available.")

    return {
        'Model': model_name,
        'Accuracy': acc,
        'Precision': prec,
        'Recall': rec,
        'F1-score': f1,
        'AUC-ROC': auc
    }


```
